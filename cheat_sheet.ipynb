# ✅ Docker CLI Cheat Sheet [Management] 👇

```bash
docker info                                      :: Display system information
docker version                                   :: Display the systems version
docker login                                     :: Log in to docker registry
```

# ✅ Docker CLI Cheat Sheet [Running and stopping] 👇

```bash
docker pull [imageName]                          :: Pull an image from a registry
docker run [imageName]                           :: Run a container
docker run -d [imageName]                        :: Detached mode
docker start [containerName]                     :: Start stopped container
docker ps                                        :: list running containers
docker ps -a                                     :: List all containers both running and stopped
docker stop [containerName]                      :: Stop containers
docker kill [containerName]                      :: Kill containers
docker image inspect [imageName]                 :: Get image details information
docker restart [containerName]                   :: Re-start containers ~
docker rm [containerName]                        :: Remove stopped containers ~ 
docker logs [containerName]                      :: View container output logs ~
```

# ✅ Docker CLI Cheat Sheet [Volumes] 👇

```bash
docker volume create [volumeName]                :: Create a new volume
docker volume ls                                 :: List all volumes
docker volume inspect [volumeName]               :: Display the volume information details
docker volume rm [volumeName]                    :: Delete a volume
docker volume prune                              :: Delete all the volumes that are not mounted
```

# ✅ Mapping a volume 👇

```bash
docker volume create myvol                                      :: Create a volume
docker volume inspect myvol                                     :: Inspect the volume
docker volume ls                                                :: List the volumes
docker run -d --name myContainer -v myvol:/app nginx:latest     :: Run a container with a volume
docker exec -it myContainer bash                                :: Connect to instance

# Creating files using nano
apt-get update
apt-get install nano
```

# ✅ YAML for Docker and Kubernates 👇 

```bash
# Comments Example

key: value
another_key: Another value
a_number_value: 10

# Nesting uses indentation. 2 space indent is preferred (not required)
a_nested_map:
    key: value
    another_key: Another Value
    another_nested_map:
        hello: Hello

# Sequences (equivalent to lists or arrays) look like this
# (NB: '-' counts as indentation):
a_sequence:
    - Item 1
    - Item 2
    - Item 3

# Json style indentaiton example (Not Convenient for YAML)
json_map: {"key": "value"}
json_sequence: [3, 2, 1, "Hello"]
```

# ✅ Docker Compose Cheat Sheet 👇 

```bash
docker compose build                                            :: Build the images
docker compose start                                            :: Start the containers
docker compose stop                                             :: Stop the containers
docker compose up -d                                            :: Build and start
docker compose ps                                               :: Lists of running containers
docker compose rm                                               :: Remove from memory
docker compose down                                             :: Stop and remove
docker compose logs                                             :: Get the logs
docker compose exec [containerName] bash                        :: Run a command in a container
```

# ✅ Docker Compose Version 2 (New Commands) Cheat Sheet 👇 

```bash
docker compose --project-name test1 up -d                       :: Run an instace as a project
docker compose -p test2 up -d                                   :: Shortcut
docker compose ls                                               :: List running projects
docker compose cp [containerID]: [SRC_PATH] [DEST_PATH]         :: Copy files from the container
docker compose cp [SRC_PATH] [containerID]: [DEST_PATH]         :: Copy files to the container
```
